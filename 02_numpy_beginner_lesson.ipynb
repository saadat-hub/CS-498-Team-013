```python
#This is an educational notebook created to teach how Python and its libraries are used in machine learning and AI.
#Using survey data from OpenAI about the use of ChatGPT we will analyze the data to extrapolate the desired information.
#The notebook will continue to use increasingly complex code to analyze the data.
```

```python
# Import the Python libraries needed for data analysis.
# Pandas organizes data into tables, while NumPy performs
# mathematical operations on collections of numbers.
from pathlib import Path
import pandas as pd
import numpy as np

# Locate the folder containing our OpenAI survey data.
folder = Path("public_release_csv")
# Read the CSV file and store its contents in a DataFrame.
# A DataFrame organizes information into rows and columns.
df = pd.read_csv(
    folder / "share_of_messages_by_topic_month.csv"
)

# Before analyzing data, we may need to prepare it.
# This process is called data preprocessing.
# Convert the month column from text into a date format.
# This allows Python to recognize and work with dates.
df["month"] = pd.to_datetime(df["month"])
# Display the first five rows to see how our data is organized.
display(df.head())
```

```python
# NumPy allows us to perform calculations on large groups
# of numbers without processing each value individually.
# Extract the message share values from our DataFrame
# and convert them into a NumPy array.
shares = df["share_of_messages"].to_numpy()
# Explore the structure of our array.
# dtype identifies the type of numbers stored.
# ndim shows how many dimensions the array has.
# shape shows how many values are stored in each dimension.
print("Data type:", shares.dtype)
print("Array dimensions:", shares.ndim)
print("Array shape:", shares.shape)
# Display only the first five values using array slicing.
print("First five values:", shares[:5])
```

```python
# Our dataset stores message shares as proportions.
# For example, 0.261 represents 26.1%.
# NumPy lets us perform calculations on every value in
# an array using a single mathematical operation.
# Multiply all proportions by 100 to obtain percentages.
percentages = shares * 100
# Compare the original values with the converted values.
print("Original proportions:", shares[:5])
print("Percentages:", percentages[:5])
```

```python
# Now that we understand arrays, we can use NumPy
# to answer questions about how people use ChatGPT.

# First, filter our dataset to include only July 2024.
july = df[df["month"] == "2024-07-01"]

# Extract the message shares into a NumPy array.
values = july["share_of_messages"].to_numpy()

# np.argmax() finds the position of the largest number.
# This helps us identify the topic with the highest share.
largest_index = np.argmax(values)

# Use that position to retrieve the corresponding topic.
top_topic = july.iloc[largest_index]["topic"]
# Display the topic and convert its share to a percentage.
print("Most common topic:", top_topic)
print("Share:", round(values[largest_index] * 100, 1), "%")
```

```python
# Select the Writing topic
writing = df[df["topic"] == "Writing"].copy()

# Sort the observations by month
writing = writing.sort_values("month")

# Convert shares into a NumPy array
writing_shares = writing["share_of_messages"].to_numpy()

# Calculate month-to-month changes
changes = np.diff(writing_shares)

print("Writing shares:", writing_shares)
print("Monthly changes:", changes)

# Find the highest and lowest monthly shares
print("Highest:", np.max(writing_shares) * 100, "%")
print("Lowest:", np.min(writing_shares) * 100, "%")
```
